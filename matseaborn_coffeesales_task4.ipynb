```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
df=pd.read_csv('Project.csv')
df
```

# **1- TEA STORE'S LOCATION BASED SALES ANALYSIS**

```python
store_loc=df['store_location'].value_counts()
store_loc
```

```python
plt.style.use('dark_background')
colors=['red','blue','green']
plt.pie(store_loc,labels=store_loc.index,colors=colors,autopct="%1.1f%%")
plt.title("LOCATION OF TEA STORES")
plt.legend(bbox_to_anchor=(1,1))
```

**INSIGHT:**
It is visible that all the three locations has almost equal sales.however,Hell's kitchen shows a comparitive hike with 34% of total sales followed by Astoria with 33.9%.

# **2- SALES BASED ON AVAILABLE PRODUCT CATEGORIES IN STORES**

```python
sns.countplot(x='product_category',data=df,palette='dark:pink')
plt.xticks(rotation=90)
plt.title('PRODUCT CATEGORY WISE SALES')
```

**INSIGHT:**
It is visible from the graph that coffee is the most selling product category after tea

# **3-LOCATION BASED PRODUCT CATEGORY SALES**

```python
sns.countplot(x='store_location',data=df,palette='flare',hue='product_category')
plt.xticks(rotation=90)
plt.title('PRODUCT CATEGORY WISE SALES')
```

**INSIGHT:**
It is visisble from the graph that Astoria is location with least selling of packaged chocolates

# **4- DIFFERENT PRODUCT TYPES PROVIDED**

```python
product_type=df["product_type"].value_counts()
product_type
```

```python
plt.figure(figsize=(10,10))
plt.bar(product_type.index,product_type.values,color="red")
plt.title("TYPES OF PRODUCTS ANALYSIS")
plt.xlabel("PRODUCT TYPE")
plt.ylabel("COUNT")
plt.xticks(rotation=90)
```

**INSIGHT:**
It is visible that Brewed chai tea is the top selling product type and Green beans the least selling type of product

# **5- MONTHLY SALES ANALYSIS**

```python
sns.countplot(x='Month Name',data=df,palette='pastel')
plt.title('MONTHLY SALES ANALYSIS')
```

**INSIGHT:**
It is can be derived from the graph that, JUNE is the season for tea sales after MAY.

# **6-SALES TRENDS OF LAST 20 DAYS**

```python
daily_sales = df.groupby(df['transaction_date']).size()
d_sales=daily_sales.sort_values(ascending=False).tail(20)
d_sales
```

```python
plt.style.use("dark_background")
plt.plot(d_sales.index,d_sales.values,"r*--")
plt.title("DAILY SALES TRENDS LAST 10 DAYS")
plt.xlabel("DATE")
plt.ylabel("TRANSACTION COUNT")
plt.xticks(rotation=90)
```

**INSIGHT:**
it can be concluded that, the trend of tea sales is little unpredictable as the graph shows an uneven hike and dip

# **7- FEBRUARY MONTH TEA SALES ANALYSIS**

```python
filt=df[df['Month Name']=='February']
feb_sale=filt['transaction_date'].value_counts().head(10)
feb_sale

```

```python
plt.bar(feb_sale.index,feb_sale.values,color='r')
plt.xticks(rotation=90)
plt.title('FEBRUARY MONTH TEA SALES ANALYSIS')
```

**INSIGHT:**
It can be derived that there is no sales hike in any particular part of the month. it is seen as any time hike or any time dip of sales

# **8- TOP 5 SELLING PRODUCT VARIENTS**

```python
plt.pie(top_varient,labels=top_varient.index,autopct="%1.1f%%")
plt.title("TOP 5 SELLING PRODUCT VARIENTS")
plt.legend(bbox_to_anchor=(1,1))
```

# **9- ANALYSIS ON SALES QUANTITY**

```python
sales_quantity=df['transaction_qty']
plt.figure(figsize=(8, 6))
sns.histplot(sales_quantity, bins=8, color='RED')
plt.title('ANALYSIS ON SALES QUANTITY')
plt.xlabel('SALES QUANTITY')
plt.ylabel('Frequency')
plt.show()
```

**INSIGHT:**
it can be concluded from the figure that sales was  happened at its top when the sales quantity is 1 while, sales has drastically fallen when the quantity of sales has increased to 3. it shows the purchase capacity of the customers that most of them are from middle class or lower middle class.



# **10- SALES TRENDS ON DIFFERENT DAYS IN A WEEK**

```python
day_sales = df.groupby(df["Day Name"]).size()
sales_trend=day_sales.sort_values()
sales_trend
```

```python
plt.style.use("dark_background")
plt.plot(sales_trend.index,sales_trend.values,marker='*',linewidth=5,markersize=10,markerfacecolor='r')
plt.title("SALES TRENDS BY DAY")
plt.xlabel("DAY")
plt.ylabel("COUNT")
plt.xticks(rotation=90)
```

**INSIGHT:**
Interestingly, its not the weekends with highest sales, but vice-versa. friday is the day that sales was at its top.
