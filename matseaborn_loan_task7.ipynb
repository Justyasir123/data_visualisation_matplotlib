```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('train.csv')
df
```

# **1-ANALYSING LOAN APPLICANTS BASED ON GENDER**

```python
gender=df['Gender'].value_counts()
gender
```

```python
plt.style.use('dark_background')
plt.pie(gender,labels=gender.index,autopct='%1.1f%%',colors=['red','blue'])
plt.title('ANALYSING LOAN APPLICANTS BASED ON GENDER')
plt.legend()
```

**INSIGHT:**
It is seen that 81.4% of loan applicants are male.

# **2- GENDER BASED INCOME STATUS OF MARRIED AND NON MARRIED APPLICANTS**

```python
sns.barplot(x=df['Gender'],y=df['ApplicantIncome'],hue='Married',data=df)
plt.title("GENDER BASED INCOME STATUS OF APPLICANTS")
```

**INSIGHT:**
It can be identified that married applicants have relatively low income than of unmarried applicants among both men and women

# **3-ANALYSIS ON GRADUATION OF APPLICANTS**

```python
edu=df['Education'].value_counts()
edu
```

```python
plt.pie(edu,labels=edu.index,autopct='%1.1f%%',colors=['blue','green'])
plt.title("ANALYSIS ON GRADUATION OF APPLICANTS")
plt.legend()
```

**INSIGHT:**
It is seen that, 78.2% of the applicants are graduates.

# **4- ANALYSIS ON LOAN AMOUNT'S DISTRIBUTION**

```python
sns.displot(df['LoanAmount'],bins=5)
plt.title("ANALYSIS ON LOAN AMOUNT'S DISTRIBUTION")
```

**INSIGHT:**
It is understood that, most of the loan amount distributed is in between 350-400.

# **5- ANALYSIS ON PROPERTY AREA AND LOAN AMOUNT**

```python
sns.lineplot(x=df['Property_Area'],y=df['LoanAmount'],data=df,errorbar=None,marker="o",markersize=15,markerfacecolor="lightgreen")
plt.title("ANALYSIS ON PROPERTY AREA AND LOAN AMOUNT")
```

**INSIGHT:**
Interestingly, properties pledged in rural areas have attained more loan amounts than of urbann or semiurban areas

# **6- LOAN APPROVAL STATUS URBAN/RURAL/SEMI URBAN PROPERTY APPLICANTS**

```python
sns.barplot(x=df['Property_Area'],y=df['LoanAmount'],hue='Loan_Status',data=df)
plt.title("GENDER BASED INCOME STATUS OF APPLICANTS")
```

**INSIGHT:**
It is visible that only urban property applicants have high loan approval status.
