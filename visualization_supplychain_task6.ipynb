```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('supply_chain_data.csv')
df
```

# **1- COMPANY LOCATION ANALYSIS**

```python
company_location=df['Location'].value_counts()
company_location
```

```python
plt.pie(company_location,labels=company_location.index,autopct="%1.1f%%")
plt.legend(title="company_location.index",bbox_to_anchor=(1.5,1))
```

**INSIGHT:**
It is seen that Kolkata with most percentage of companies with 25%

# **2- MEANS OF TRANSPORTATION ROUTES USED FOR SUPPLY CHAIN OF MAKEUP PRODUCTS**

```python
Trans=df["Transportation modes"].value_counts()
Trans
```

```python
plt.style.use('dark_background')
plt.pie(Trans,labels=Trans.index,autopct="%1.1f%%",colors=["red","yellow","green","violet"])
plt.legend(title="Trans.index",bbox_to_anchor=(1.5,1))
plt.title("Transportation")
```

**INSIGHT:**
It can be analysed that, road is mostly used means of transportation among all others with 29%.

# **3- ANALYSIS ON PRODUCTS AND THEIR AVAILABILITY**

```python
sns.barplot(x=df["Product type"],y=df["Availability"],data=df,color="lightblue")
plt.title("Products and their Availability")
```

**INSIGHT:**
It is seen that cosmetic products are comparitively high in stock.

# **4- ANALYSIS ON PRODUCTS AND THEIR RESPECTIVE SALES**

```python
sns.barplot(x=df["Product type"],y=df["Number of products sold"],data=df,color="brown")
plt.title("Products Vs No Of Product Sold")
```

**INSIGHT:**
It can be understood that skin care products are gaining high sales

# **5- ANALYSIS ON SHIPPING TIME DISTRIBUTION**

```python
sns.displot(df['Shipping times'],bins=8)
plt.title("SHIPPING TIME'S DISTRIBUTION")
```

**INSIGHT:**
It can be derived that most of the products are taking a shipping time of 9-10 days to be delivered.

# **6- ANALYSIS ON DIFFERENT PRODUCT'S REVENUE DISTRIBUTION**

```python
sns.lineplot(x=df["Product type"],y=df["Revenue generated"],data=df,errorbar=None,marker="o",markersize=15,markerfacecolor="lightgreen")
plt.title("Products Vs Revenue")
```

**INSIGHT:**
cosmetic products have gained highest revenue

# **7- ANALYSIS ON PRODUCT INSPECTION STATUS BEFORE SUPPLY**

```python
inspection=df["Inspection results"].value_counts()
inspection
```

```python
plt.pie(inspection,labels=inspection.index,autopct="%1.1f%%",colors=["lightgreen","indigo","purple"])
plt.legend(title="inspection.index",bbox_to_anchor=(1.5,1))
```

**INSIGHT:**
It is seen that 41% of product's inspection is pending and 36% of product's inspection status if failed and 23% of product's inspection is only passed.

# **8- ANALYSIS ON HAIRCARE PRODUCT'S CUSTOMER DEMOGRAPHICS BASED ON QUANTITY**

```python
filter1=df[df["Product type"]=="haircare"]
sns.violinplot(x="Customer demographics",y="Order quantities",data=filter1)
plt.title("Haircare Product's Order Quantity Wise Customer Demographics")
```

**INSIGHT:**
It is seen that, non binary customers are ordering high quantity of hair care products

# **9- ANALYSIS ON PRODUCTS AND THEIR MANUFACTURING COSTS**

```python
sns.barplot(x="Product type",y="Manufacturing costs",data=df,color="green")
plt.title("Products and their Manufacturing Costs")
```

**INSIGHT:**
It is visible that cosmetic products have less manufacturing costs

# **10-DEFECT RATES OF PRODUCTS EVEN IF THEIR INSPECTION STATUS IS PASS**

```python
filter2=df[df["Inspection results"]=="Pass"]
sns.lineplot(x="Product type",y="Defect rates",data=filter2,marker="o",errorbar=None,markersize=15,markerfacecolor="indigo")
plt.title('DEFECT RATES OF PRODUCTS EVEN IF THEIR INSPECTION STATUS IS PASS')
```

**INSIGHT:**
Hair care products are having high defect rates even if their inspection status is passed

# **11- DETAILED ANALYSIS ON PRODUCTS AND AVAILABLITY ALONG WITH THEIR INSPECTION STATUS**

```python
sns.barplot(x="Product type",y="Stock levels",hue="Inspection results",data=df,palette='flare')
plt.title("Inspection Results Passed Products And Their Stock Levels")
```

**INSIGHT:**
It is seen that, cosmetic products are having higher stock levels with inspection status 'pending' and 'passed'.

# **16- DEFECTS RATE BASED ON DIFFERENT MODES OF TRANSPORTATION**

```python
defect_rate_trans=df.groupby("Transportation modes")["Defect rates"].mean()
```

```python
plt.style.use("bmh")
plt.pie(defect_rate_trans,labels=defect_rate_trans.index,autopct="%1.1f%%")
```

**INSIGHT:**
It is visible from the graph Air transportation is showing least defect rates during supply chain

# **17- ANALYSING LEAD TIMES BY PRODUCT TYPE**

```python
plt.figure(figsize=(8,6))
sns.boxplot(data=df, x='Product type', y='Lead times',palette="pastel")
plt.xlabel('Product Type')
plt.ylabel('Lead Times')
plt.title('Lead Times by Product Type')
```

**INSIGHT:**
NO outlayers are found

# **18- AVERAGE LEAD TIME BY TRANSPORTATION MODE**

```python
trans_lead=df.groupby(['Transportation modes'])['Lead times'].mean()
```

```python
plt.bar(trans_lead.index,trans_lead.values,color=["brown","silver","grey","gold"])
plt.title("AVERAGE LEAD TIME BY TRANSPORTATION MODE")
plt.xlabel("TRANSPORTATION MODE")
plt.ylabel("AVERAGE LEAD TIME")
```

**INSIGHT:**
Ait trasnsport is showing highest lead time followed by road transport

# **19-FINDING AVERAGE LEAD TIMES BY DIFFERENT ROUTES**

```python
route_lead=df.groupby('Routes')['Lead times'].mean()
route_lead
```

```python
plt.pie(route_lead,labels=route_lead.index,autopct="%1.1f%%",explode=[0,.1,0])
plt.title("AVERAGE LEAD TIMES BY ROUTE")
```

**INSIGHT:**
ROUTE B has highest average lead time with 35.7%.
